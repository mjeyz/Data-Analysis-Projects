```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("healthcare_patient_analytics_seaborn.csv")
```

```python
df.head()
```

```python
sns.scatterplot(data=df, x="treatment_cost", y="recovery_score", hue="visit_type")
```

```python
sns.relplot(data=df, x="treatment_cost", y="recovery_score", hue="visit_type", col="gender")
```

# Difference between Seaborn and Matplotlib:
Seaborn is build on top of Matplotlib and provide a higher level interface for creating more attractive and informative statistical graphics.
like above grove plot, it is more informative than the simple scatter plot created by Seaborn but Chart created by Matplotlib is less attractive lengthy and confusing because it requires more code to create the same plot and it does not have built-in support for handling categorical variables and color coding based on categories.

```python
plt.plot(df["treatment_cost"], df["recovery_score"], "o")
plt.xlabel("Treatment cost")
plt.ylabel("Recovery score")
plt.show()
```

```python
sns.lineplot(data=df, x="treatment_cost", y="recovery_score", hue="visit_type")
```

```python
sns.histplot(data=df, x="readmission_risk", hue="visit_type", bins=30)
plt.title("Histogram of Readmission Risk by Visit Type")
```

```python
sns.kdeplot(data=df, x="recovery_score", hue="visit_type", fill=True)
```

```python
sns.ecdfplot(data=df, x="length_of_stay_days", hue="visit_type")
plt.title("ECDF of Length of Stay by Visit Type")
```

```python
sns.violinplot(data=df, x="department", y="recovery_score")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.stripplot(data=df, x="department", y="recovery_score", jitter=True)
plt.xticks(rotation=45)
plt.title("Strip plot of Department and Recovery Score")
plt.show()
```

```python
sns.boxplot(data=df, x="department", y="recovery_score")
plt.xticks(rotation=45)
plt.title("Box plot of Department and Recovery Score")
plt.show()
```
